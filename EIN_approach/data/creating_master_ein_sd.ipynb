```python
import pandas as pd
pd.set_option('display.max_columns', None)
pp = pd.read_csv('data/propublica_SD_located_nonprofits.csv')

```

```python
irs_eo = pd.read_csv('data/irs_eo_san_diego.csv')
irs_eo = irs_eo[irs_eo['CITY']=='SAN DIEGO']
```

Checking lengths of EINs before merging again

```python
print(irs_eo["EIN"].astype(str).str.len().value_counts().sort_index())
print(pp["ein"].astype(str).str.len().value_counts().sort_index())
irs_eo['EIN'] = irs_eo['EIN'].astype(str).str.zfill(9)
pp['EIN'] = pp['ein'].astype(str).str.zfill(9)
merged_ein = irs_eo.merge(pp, on = 'EIN', how = 'outer', indicator=True)
merged_ein
```

```python
print(merged_ein["_merge"].value_counts())
print("matched %:", (merged_ein["_merge"] == "both").mean() * 100)
```

```python
merged_ein.to_csv('Master_ein_sd.csv', index = False)
```
