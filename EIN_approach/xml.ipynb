```python
import pandas as pd
xml = pd.read_csv('xml_people_df.csv').dropna(subset=['ein', 'website', 'person_name'])
len(xml['org_name'].unique())
```

```python
len(xml['ein'].unique())
```

```python
def clean_website(website):
    website = str(website).lower().strip()
    for prefix in ['https://', 'http://']:
      if website.startswith(prefix):
        website = website[len(prefix):]
    if website.startswith('www.'):
        website = website[4:]
    website = website.split("/")[0]
    website = website.strip('.')

    return website

xml['domain'] = xml['website'].apply(clean_website)
xml['person_name'] = xml['person_name'].str.lower()
name_parts = xml['person_name'].str.split()
xml['first'] = name_parts.str[0]
xml['last'] = name_parts.str[-1]
```

```python
xml['guess1'] = xml['first'] + xml['domain']
xml['guess2'] = xml['first'] + "." + xml['last'] + xml['domain']
xml['guess3'] = xml['first'][0] + xml['last'] + xml['domain']
xml['guess4'] = xml['first'] + xml['last'] + xml['domain']
```

```python
xml.to_csv('XML_EIN_emails.csv')
```
